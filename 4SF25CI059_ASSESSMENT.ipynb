```python
import random


ranks = ['2', '3', '4', '5', '6', '7', '8', '9', '10', 'Jack', 'Queen', 'King', 'Ace']
suits = ['Hearts', 'Diamonds', 'Clubs', 'Spades']

deck = [f"{rank} of {suit}" for suit in suits for rank in ranks]


random.shuffle(deck)


while True:
    try:
        num_players = int(input("Enter number of players: "))
        if num_players <= 0:
            print("Error: Please enter a positive number greater than 0.")
        elif num_players > 52:
            print("Error: Player count cannot be more than the total cards (52).")
        else:
            break
    except ValueError:
        print("Error: Invalid input! Please enter a whole number.")


cards_per_player = len(deck) // num_players
total_rounds = cards_per_player

hands = []
for i in range(num_players):
    start = i * cards_per_player
    end = start + cards_per_player
    hands.append(deck[start:end])


scores = [0] * num_players

print(f"\nGame Started! Each player gets {cards_per_player} cards.")
print(f"Total rounds to be played: {total_rounds}\n")


for r in range(1, total_rounds + 1):
    print(f"--- Round {r} ---")
    
    
    for i in range(num_players):
        selected_card = random.choice(hands[i])
        hands[i].remove(selected_card)
        print(f"Player {i + 1} played: {selected_card}")
    
    
    while True:
        try:
            winner = int(input(f"Enter the winner of this round (1 to {num_players}): "))
            if 1 <= winner <= num_players:
                scores[winner - 1] += 1
                break
            else:
                print(f"Error: Enter a valid player number between 1 and {num_players}.")
        except ValueError:
            print("Error: Please enter a valid number.")
    print()


print("--- Final Scores ---")
for i in range(num_players):
    print(f"Player {i + 1}: {scores[i]} points")

highest_score = max(scores)
overall_winners = []
for i in range(num_players):
    if scores[i] == highest_score:
        overall_winners.append(i + 1)

print("\n--- Result ---")
if len(overall_winners) == 1:
    print(f"Player {overall_winners[0]} wins the game!")
else:
    print(f"It's a tie between players: {overall_winners}")
```
